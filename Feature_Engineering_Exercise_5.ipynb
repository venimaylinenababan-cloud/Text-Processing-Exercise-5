```python
import pandas as pd
df = pd.read_csv('candy_crush_preprocessed (2).csv')
df = df.dropna().reset_index().drop(columns=['index'])
df_pre = df['stem']
df_pre.info()
```

```python
import warnings
warnings.filterwarnings(action='ignore')
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Mengubah teks menjadi matriks TF-IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df_pre)
df_FE = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
df_FE
```

Tokenize Sentence

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk import word_tokenize
from tqdm import tqdm

sentences = [word_tokenize(tweet.lower()) for tweet in tqdm(df_pre)]
sentences
```

Build Word2Vec moldel


```python
!pip install gensim
```

```python
import gensim.models
# 1. Train Model
model = gensim.models.Word2Vec(sentences, min_count=1, vector_size=100, window=5, sg=0)
```

```python
# 2. Save Model
model.save('sentimen_tweet.w2v')
```

```python
# 3. Load Model
model = gensim.models.Word2Vec.load('sentimen_tweet.w2v')
```

```python
# 4. Model Information
w2v = model.wv
w2v.index_to_key[:10]
```

```python
w2v.vectors.shape, w2v.vector_size # size of vector
```

```python
w2v.similar_by_key('game')
```

Visualize


```python
!pip install umap-learn
```

```python
from umap import UMAP
```

```python
import numpy as np
import plotly.express as px
```

```python
# Tentukan jumlah kata teratas yang mau ditampilkan (misal 50 kata)
top_n = 50

# Mengambil vektor dan kata teratas saja
vectors_top = w2v.vectors[:top_n]
words_top = w2v.index_to_key[:top_n]

# Reduksi dimensi dengan UMAP
X = UMAP().fit_transform(vectors_top)

df2 = pd.DataFrame(X, columns=['umap1', 'umap2'])
df2['word'] = words_top

# Visualisasi Plotly
fig = px.scatter(df2, x='umap1', y='umap2', text='word')
fig.update_traces(textposition='top center')
fig.update_layout(height=800, title_text=f'Word2Vec Visualization (Top {top_n} Words)')
fig.show()
```

## **Build FastText Model**

```python
import gensim.models
import random
import numpy as np
#build model
model_fasttext = gensim.models.FastText(sentences, min_count=1,
                                        vector_size=100, window=5,
                                        min_n=3, max_n=6,
                                        sg=1, epochs=10)
```

```python
seed_value = 42
random.seed(seed_value)
np.random.seed(seed_value)
```

```python
model_fasttext = gensim.models.FastText(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    min_n=3,
    max_n=6,
    sg=1,
    epochs=10,
    workers=1,      # Mengunci multi-threading
    seed=seed_value # Mengunci bobot acak
)
```

```python
# 2. save model
model_fasttext.save('sentimen_tweet.fasttext')
```

```python
# 3. load model
model_fasttext = gensim.models.Word2Vec.load('sentimen_tweet.fasttext')
```

```python
# 4. model information
fastext = model_fasttext.wv
fastext.index_to_key[:10]
```

```python
fastext.vectors.shape
```

```python
fastext.similar_by_key('4g')
```

```python
# 5. FE result
import numpy as np
def get_document_vector(tokens, model):
    valid_vectors = [model.wv[word] for word in tokens if word in model.wv]
    # if not valid_vectors:
    #   return np.zeros(model.vector_size)
    return np.mean(valid_vectors, axis=0)
X_features = np.array([get_document_vector(doc, model_fasttext) for doc in sentences])
pd.DataFrame(X_features)
```

```python
top_n = 50

vectors_top = fastext.vectors[:top_n]
words_top = fastext.index_to_key[:top_n]

X = UMAP(random_state=42).fit_transform(vectors_top)

df2 = pd.DataFrame(X, columns=['umap1', 'umap2'])
df2['word'] = words_top

fig = px.scatter(df2, x='umap1', y='umap2', text='word')
fig.update_traces(textposition='top center')
fig.update_layout(height=800, title_text=f'FastText Visualization (Top {top_n} Words)')
```
